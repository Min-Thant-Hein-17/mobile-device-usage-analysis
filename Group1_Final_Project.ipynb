# Exploring whether age is related to the number of apps installed and screen time

````python
# Categorize age into groups: Young (< 30) and Old (≥ 30)
device_df['Age Group'] = device_df['Age'].apply(lambda x: 'Young' if x < 30 else 'Old')

# Group by Age Group to calculate average screen time and apps installed
age_group_analysis = device_df.groupby('Age Group').agg({
    'Screen On Time (hours/day)': 'mean',
    'Number of Apps Installed': 'mean'
}).reset_index()

````

````python
# Visualization: Screen On Time by Age Group with Exact Hours
plt.figure(figsize=(8, 5))
barplot = sns.barplot(x='Age Group', y='Screen On Time (hours/day)', data=age_group_analysis, palette='Blues')

# Add exact values on top of the bars
for bar in barplot.patches:
    # Get the x and y positions of each bar
    x = bar.get_x() + bar.get_width() / 2
    y = bar.get_height()
    # Annotate the bar with the value (rounded to 1 decimal place)
    plt.text(x, y, f'{y:.1f}', ha='center', va='bottom', fontsize=10, color='black')

# Add labels, title, and formatting
plt.title('Average Screen Time by Age Group', fontsize=14)
plt.ylabel('Screen On Time (hours/day)', fontsize=12)
plt.xlabel('Age Group', fontsize=12)
plt.tight_layout()
plt.show()
````

````python
# Visualization: Number of Apps Installed by Age Group with Exact Numbers
plt.figure(figsize=(8, 5))
barplot = sns.barplot(x='Age Group', y='Number of Apps Installed', data=age_group_analysis, palette='Greens')

# Add exact values on top of the bars
for bar in barplot.patches:
    # Get the x and y positions of each bar
    x = bar.get_x() + bar.get_width() / 2
    y = bar.get_height()
    # Annotate the bar with the value (rounded to 0 decimal place)
    plt.text(x, y, f'{y:.0f}', ha='center', va='bottom', fontsize=10, color='black')

# Add labels, title, and formatting
plt.title('Average Number of Apps Installed by Age Group', fontsize=14)
plt.ylabel('Number of Apps Installed', fontsize=12)
plt.xlabel('Age Group', fontsize=12)
plt.tight_layout()
plt.show()
````
